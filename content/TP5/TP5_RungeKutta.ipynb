# TP 5 — Les méthodes de Runge–Kutta

**L3 Mathématiques · Équations différentielles et optimisation · Université d'Artois · 2026–2027**

Au TP 3, la méthode de Taylor d'ordre $n$ atteignait l'ordre $n$… à condition de calculer les dérivées totales
$f', f'', \dots$, à la main ou avec SymPy, et leur taille explosait. Les **méthodes de Runge–Kutta** atteignent le
même ordre en n'évaluant **que $f$**, en plusieurs points bien choisis à l'intérieur de chaque pas. Dans ce TP, vous
allez :

- programmer les méthodes de Runge–Kutta d'ordre 2 (point milieu, Euler modifiée…) et vérifier avec SymPy
  **pourquoi** elles sont d'ordre 2 ;
- programmer la méthode de Runge–Kutta « classique » d'ordre 4 (**RK4**) et mesurer son ordre ;
- décrire une méthode par son **tableau de Butcher**, et lire son ordre directement sur ce tableau ;
- comparer les méthodes **à coût égal** (même nombre d'évaluations de $f$) ;
- appliquer RK4 à un **système** d'EDO, issu d'une équation d'ordre 2 ;
- ★ étudier la stabilité de RK4, puis un modèle proies–prédateurs.

### Rappels 📖

On considère le problème de Cauchy

$$y'(t) = f\big(t, y(t)\big), \quad t \in [t_0, T], \qquad y(t_0) = y_0,$$

et la grille $t_i = t_0 + i\,h$, $i = 0, \dots, N$, de pas $h = \dfrac{T - t_0}{N}$.

**Méthodes à un pas** (TP 3). À partir de $w_0 = y_0$, on calcule $w_{i+1} = w_i + h\, \varphi(t_i, w_i, h)$ :

| Méthode | Pente $\varphi(t, w, h)$ | Évaluations par pas | Ordre |
|:--|:--|:--|:--|
| Euler | $f(t, w)$ | $1$ : $f$ | 1 |
| Euler modifiée (TP 2) | $\frac{1}{2}\big[f(t, w) + f\big(t + h,\ w + h\, f(t, w)\big)\big]$ | $2$ : $f$ deux fois | 2 |
| Taylor d'ordre $n$ (TP 3) | $T^{(n)} = f + \frac{h}{2}\, f' + \dots + \frac{h^{n-1}}{n!}\, f^{(n-1)}$ | $n$ : $f, f', \dots, f^{(n-1)}$ | $n$ |

Si l'**erreur de consistance** $\tau_{i+1}(h) = \dfrac{y(t_{i+1}) - y(t_i)}{h} - \varphi\big(t_i, y(t_i), h\big)$
est en $O(h^p)$ et si $\varphi$ est lipschitzienne en $w$, l'erreur globale est en $O(h^p)$ : la méthode est
**d'ordre $p$** (théorème du TP 3).

**Le défaut des méthodes de Taylor** : il faut calculer les dérivées totales $f' = \partial_t f + f\,\partial_y f$,
$f''$, … qui deviennent vite énormes. **L'idée de Runge–Kutta** : remplacer ces dérivées par des **évaluations de
$f$ en des points intermédiaires** du pas, bien choisis. On garde l'ordre de Taylor, sans rien dériver.

**Exemple fil rouge** (TP 2 et 3) :

$$y' = y - t^2 + 1, \quad t \in [0, 2], \quad y(0) = 0{,}5, \qquad \text{de solution exacte }
y(t) = (t + 1)^2 - \tfrac{1}{2}\, e^{t}.$$

La cellule suivante définit $f$, la solution exacte et la méthode d'Euler du TP 2 (pour les comparaisons) 📖 :

```python
def f(t, y):
    """Second membre de l'exemple fil rouge : y' = y - t^2 + 1."""
    return y - t**2 + 1


def y_exacte(t):
    """Solution exacte de l'exemple fil rouge."""
    return (t + 1)**2 - 0.5 * np.exp(t)


def euler(f, t0, T, y0, N):
    """Méthode d'Euler (TP 2), fournie pour les comparaisons."""
    t = np.linspace(t0, T, N + 1)
    h = (T - t0) / N
    w = np.zeros(N + 1)
    w[0] = y0
    for i in range(N):
        w[i + 1] = w[i] + h * f(t[i], w[i])
    return t, w
```

## 1. Les méthodes de Runge–Kutta d'ordre 2

### L'idée de Runge 📖

La méthode de Taylor d'ordre 2 utilise la pente

$$T^{(2)}(t, w) = f(t, w) + \frac{h}{2}\, f'(t, w) = f + \frac{h}{2}\, \partial_t f + \frac{h}{2}\, f\, \partial_y f
\qquad \text{(tout est évalué en } (t, w)\text{)}.$$

Or la **formule de Taylor à deux variables** donne, pour $f$ assez régulière :

$$f\Big(t + \frac{h}{2},\ w + \frac{h}{2}\, f(t, w)\Big) = f + \frac{h}{2}\, \partial_t f + \frac{h}{2}\, f\, \partial_y f
+ O(h^2) = T^{(2)}(t, w) + O(h^2).$$

Une **seule** évaluation de $f$, au bon endroit, remplace $f$ et $f'$ ! C'est la **méthode du point milieu** :

$$\boxed{\; k_1 = f(t_i, w_i), \qquad k_2 = f\Big(t_i + \frac{h}{2},\ w_i + \frac{h}{2}\, k_1\Big), \qquad
w_{i+1} = w_i + h\, k_2. \;}$$

$k_1$ est la pente au début du pas ; un demi-pas d'Euler donne un point « au milieu » du pas, où l'on calcule la
pente $k_2$ ; on fait alors tout le pas avec la pente $k_2$.

## 2. La méthode de Runge–Kutta classique (RK4)

### Rappel 📖

Avec **quatre** évaluations de $f$ par pas, on atteint l'ordre 4 :

$$\boxed{\begin{aligned}
k_1 &= f(t_i, w_i), & k_2 &= f\Big(t_i + \frac{h}{2},\ w_i + \frac{h}{2}\, k_1\Big), \\
k_3 &= f\Big(t_i + \frac{h}{2},\ w_i + \frac{h}{2}\, k_2\Big), & k_4 &= f(t_i + h,\ w_i + h\, k_3), \\
w_{i+1} &= w_i + \frac{h}{6}\, (k_1 + 2 k_2 + 2 k_3 + k_4). & &
\end{aligned}}$$

$k_1$ est la pente au début du pas, $k_2$ et $k_3$ deux estimations de la pente au milieu, $k_4$ une estimation de
la pente à la fin. Les poids $\frac{1}{6}, \frac{2}{6}, \frac{2}{6}, \frac{1}{6}$ rappellent la formule de Simpson
$\int_0^1 g \approx \frac{1}{6}\big[g(0) + 4\, g(\tfrac{1}{2}) + g(1)\big]$. C'est la méthode de Runge–Kutta la plus
utilisée ; elle est d'ordre 4 si $y$ est de classe $\mathcal{C}^5$.

> Burden & Faires écrivent $k_1 = h f(t_i, w_i)$, …, $w_{i+1} = w_i + \frac{1}{6}(k_1 + 2k_2 + 2k_3 + k_4)$ : leurs
> $k_j$ sont nos $h\, k_j$. Le résultat est le même.

## 3. Tableaux de Butcher et conditions d'ordre

### Rappel 📖

Une méthode de Runge–Kutta **explicite à $s$ étages** est définie par des coefficients $c_j$, $a_{jl}$ ($l < j$) et
$b_j$ :

$$k_j = f\Big(t_i + c_j h,\ w_i + h \sum_{l < j} a_{jl}\, k_l\Big), \quad j = 1, \dots, s, \qquad
w_{i+1} = w_i + h \sum_{j=1}^{s} b_j\, k_j.$$

On les range dans le **tableau de Butcher** ($c_1 = 0$, et $c_j = \sum_l a_{jl}$) :

$$\begin{array}{c|cccc}
c_1 & & & & \\
c_2 & a_{21} & & & \\
\vdots & \vdots & \ddots & & \\
c_s & a_{s1} & \cdots & a_{s,s-1} & \\ \hline
 & b_1 & \cdots & b_{s-1} & b_s
\end{array}
\qquad \text{par exemple} \qquad
\underset{\text{point milieu}}{\begin{array}{c|cc}
0 & & \\
\frac{1}{2} & \frac{1}{2} & \\ \hline
 & 0 & 1
\end{array}}
\qquad
\underset{\text{RK4}}{\begin{array}{c|cccc}
0 & & & & \\
\frac{1}{2} & \frac{1}{2} & & & \\
\frac{1}{2} & 0 & \frac{1}{2} & & \\
1 & 0 & 0 & 1 & \\ \hline
 & \frac{1}{6} & \frac{1}{3} & \frac{1}{3} & \frac{1}{6}
\end{array}}$$

En Python, on stocke `A` (tableau $s \times s$, nul sur et au-dessus de la diagonale), `b` et `c`.

### Les conditions d'ordre 📖

**Théorème (admis ; Butcher).** Une méthode de Runge–Kutta explicite, avec $c_j = \sum_l a_{jl}$, est d'ordre
$\geq p$ si et seulement si elle vérifie toutes les conditions des lignes $1$ à $p$ du tableau suivant (sommes sur
$i, j, k$ de $1$ à $s$) :

| Ordre | Conditions | En NumPy |
|:--|:--|:--|
| 1 | $\sum_i b_i = 1$ | `b.sum()` |
| 2 | $\sum_i b_i c_i = \frac{1}{2}$ | `b @ c` |
| 3 | $\sum_i b_i c_i^2 = \frac{1}{3}$, $\quad \sum_{i,j} b_i a_{ij} c_j = \frac{1}{6}$ | `b @ c**2`, `b @ A @ c` |
| 4 | $\sum_i b_i c_i^3 = \frac{1}{4}$, $\quad \sum_{i,j} b_i c_i a_{ij} c_j = \frac{1}{8}$, $\quad \sum_{i,j} b_i a_{ij} c_j^2 = \frac{1}{12}$, $\quad \sum_{i,j,k} b_i a_{ij} a_{jk} c_k = \frac{1}{24}$ | `b @ c**3`, `(b * c) @ A @ c`, `b @ A @ c**2`, `b @ A @ A @ c` |

La ligne 1 est la consistance ; la ligne 2 est la condition $b_2 \alpha = \frac{1}{2}$ des méthodes RK2.

#### ✏️ Exercice 7 — Lire l'ordre sur le tableau

Compléter `ordre_tableau(A, b, c)`, qui renvoie le plus grand $p \leq 4$ tel que les conditions des lignes $1$ à

moins membre de droite » d'une ligne du tableau ; elles doivent être nulles.

*Indication* : `np.allclose(liste, 0)` teste si tous les nombres de `liste` sont nuls (aux erreurs d'arrondi près).

```python
def ordre_tableau(A, b, c):
    """Plus grand p <= 4 tel que la méthode de tableau (A, b, c) vérifie les conditions d'ordre 1, ..., p."""
    A, b, c = np.array(A, dtype=float), np.array(b, dtype=float), np.array(c, dtype=float)
    conditions = [
        [b.sum() - 1],                           # ordre 1
        [b @ c - 1 / 2],                         # ordre 2
        [..., ...],                              # ordre 3
        [..., ..., ..., ...],                    # ordre 4
    ]
    p = 0
    for liste in conditions:
        if ...:                                  # toutes les conditions de la liste sont-elles vérifiées ?
            p = p + 1
        else:
            break
    return p
```

## 7. ★ Proies et prédateurs : une quantité conservée

On reprend le modèle de Lotka–Volterra (TP 1), avec les données de Burden & Faires (§ 5.9, exercice 9) : $x_1(t)$
proies et $x_2(t)$ prédateurs,

$$x_1' = k_1 x_1 - k_2 x_1 x_2, \qquad x_2' = k_3 x_1 x_2 - k_4 x_2,$$

avec $k_1 = 3$, $k_2 = 0{,}002$, $k_3 = 0{,}0006$, $k_4 = 0{,}5$, $x_1(0) = 1000$, $x_2(0) = 500$.

La quantité $H(x_1, x_2) = k_3 x_1 - k_4 \ln x_1 + k_2 x_2 - k_1 \ln x_2$ est **constante** le long des solutions
exactes : les trajectoires sont des courbes fermées. Une bonne méthode numérique doit (presque) conserver $H$. La
cellule suivante définit les constantes, $H$ et la méthode d'Euler pour les systèmes du TP 2 📖 :

```python
K1, K2, K3, K4 = 3, 0.002, 0.0006, 0.5


def H(X):
    """Quantité conservée ; X est un vecteur (x1, x2) ou un tableau dont chaque ligne est un tel vecteur."""
    X = np.asarray(X, dtype=float)
    return K3 * X[..., 0] - K4 * np.log(X[..., 0]) + K2 * X[..., 1] - K1 * np.log(X[..., 1])


def euler_systeme(F, t0, T, Y0, N):
    """Méthode d'Euler pour un système (TP 2), fournie pour les comparaisons."""
    t = np.linspace(t0, T, N + 1)
    h = (T - t0) / N
    W = np.zeros((N + 1, len(Y0)))
    W[0] = Y0
    for i in range(N):
        W[i + 1] = W[i] + h * F(t[i], W[i])
    return t, W
```

## 8. Aide-mémoire

| Besoin | Commande |
|:--|:--|
| Pente au milieu du pas | `f(t[i] + h / 2, w[i] + h / 2 * k1)` |
| Produit scalaire, produit matrice-vecteur | `b @ k`, `A @ c`, `b @ A @ c` |
| Les $j$ premiers éléments, ligne $j$ de `A` | `k[:j]`, `A[j, :j]` |
| Tableau de vecteurs (systèmes) | `W = np.zeros((N + 1, m))` ; la ligne `W[i]` est le vecteur $W_i$ |
| Lignes $10, 20, \dots$ d'un tableau | `X[[10, 20, 30, 40]]` |
| Tous les nombres sont nuls (aux arrondis près) | `np.allclose(liste, 0)` |
| Rapports successifs, ordre | `E[:, :-1] / E[:, 1:]`, `np.log2(...)` |
| Fonction inconnue (SymPy) | `fs = sp.Function("f")` ; `fs(ts, ys)` |
| Développement limité en $h$ | `sp.series(expr, hs, 0, 2).removeO()` |

### À retenir

- **Runge–Kutta** : on remplace les dérivées de $f$ (Taylor) par des évaluations de $f$ en des points intermédiaires
  du pas. Les méthodes RK2 (point milieu, Euler modifiée…) sont d'ordre 2, RK4 est d'ordre 4.
- Une méthode explicite est décrite par son **tableau de Butcher** $(c, A, b)$ ; son ordre se lit sur des
  **conditions algébriques** : $\sum b_i = 1$, $\sum b_i c_i = \frac{1}{2}$, …
- Le coût se mesure en **évaluations de $f$** : à coût égal, une méthode d'ordre élevé est bien plus précise.
- Les mêmes formules s'appliquent aux **systèmes** (vecteurs $k_j$), donc aux équations d'ordre supérieur.

> **Pour aller plus loin.** La fonction `solve_ivp` de SciPy utilise par défaut une méthode de Runge–Kutta d'ordre 5
> (`method="RK45"`, de Dormand et Prince), qui estime aussi son erreur à chaque pas avec une méthode d'ordre 4
> (méthode « emboîtée ») et **adapte le pas** en conséquence.
