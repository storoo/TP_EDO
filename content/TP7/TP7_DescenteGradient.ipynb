# TP 7 — La méthode du gradient

**L3 Mathématiques · Équations différentielles et optimisation · Université d'Artois · 2026–2027**

Au TP 6, on a **reconnu** les minima d'une fonction : ce sont des points critiques, où la dérivée (le gradient)
s'annule. Il reste à les **trouver**. La **méthode du gradient** part d'un point et descend la pente, pas à pas :
chercher le sens dans lequel la fonction descend, faire un pas dans ce sens, recommencer. Proposée par Cauchy en
1847, c'est encore elle qui, sous le nom de *rétropropagation*, entraîne les réseaux de neurones.

Dans ce TP, vous allez :

- revoir la **dérivée** comme pente de la tangente, limite des pentes des sécantes, et la calculer numériquement ;

- l'appliquer à la **régression linéaire** (méthode des moindres carrés), sur des données que vous fabriquerez ;
- l'améliorer avec un **moment**, qui donne de l'élan à la descente ;
- ★ passer à deux paramètres, avec le gradient du TP 6.

```python
# 🧰 Boîte à outils du TP 7 — exécutée automatiquement à l'ouverture (sinon : Maj + Entrée).
# tracer_iteres(f, points, xlim, ax=ax) : graphe de f et points (a_k, f(a_k)) d'une méthode, renvoie ax ;
# tracer_contours(f, xlim, ylim, ax=ax) : lignes de niveau de f : R^2 -> R, renvoie ax (TP 6) ;
# grad_num(f, x) : gradient par différences centrées (TP 6).
import math
import numpy as np
import matplotlib.pyplot as plt

def tracer_contours(f, xlim=(-2, 2), ylim=(-2, 2), n=200, niveaux=20, ax=None):
    """Trace les lignes de niveau de f : R^2 -> R sur le rectangle xlim × ylim et renvoie ax.

    f prend un point x = np.array([x, y]) et renvoie un nombre. niveaux : nombre de lignes de niveau
    (réparties selon les valeurs de f) ou liste des valeurs voulues."""
    if ax is None:
        fig, ax = plt.subplots(figsize=(5.5, 5))
    X, Y = np.meshgrid(np.linspace(xlim[0], xlim[1], n), np.linspace(ylim[0], ylim[1], n))
    try:
        Z = np.asarray(f(np.array([X, Y])), dtype=float)        # rapide si f est écrite avec x[0] et x[1]
        if Z.shape != X.shape:
            raise ValueError
    except Exception:                                           # sinon, point par point
        Z = np.array([[f(np.array([a, b])) for a, b in zip(ligne_x, ligne_y)]
                      for ligne_x, ligne_y in zip(X, Y)], dtype=float)
    if np.isscalar(niveaux):
        niveaux = np.unique(np.quantile(Z, np.linspace(0, 1, int(niveaux) + 2)[1:-1]))
    ax.contour(X, Y, Z, levels=niveaux, cmap="viridis", linewidths=1)
    ax.set_xlabel("$x$")
    ax.set_ylabel("$y$")
    ax.set_aspect("equal")
    return ax

def grad_num(f, x, h=1e-6):
    """Gradient de f en x par différences centrées (TP 6) : (f(x + h e_i) - f(x - h e_i)) / (2h).

    Renvoie un tableau de forme (n,) ; x n'est pas modifié. Erreur en O(h^2) : environ 1e-10 avec h = 1e-6."""
    x = np.asarray(x, dtype=float)
    n = len(x)
    g = np.zeros(n)
    for i in range(n):
        e = np.zeros(n)
        e[i] = 1.0                       # i-ème vecteur de la base canonique
        g[i] = (f(x + h * e) - f(x - h * e)) / (2 * h)
    return g

def tracer_iteres(f, points, xlim, ax=None, couleur="C3", nom="itérés"):
    """Trace le graphe de f sur l'intervalle xlim, puis les points (a_k, f(a_k)) d'une méthode itérative,
    reliés dans l'ordre (TP 7). Le point de départ est entouré, le dernier point est plus gros. Renvoie ax.

    f doit s'appliquer à un tableau (np.sin et non math.sin) ; points : liste ou tableau des a_k.
    Le cadre reste celui de xlim : les points qui en sortent (méthode qui diverge) ne sont pas visibles."""
    if ax is None:
        fig, ax = plt.subplots(figsize=(7, 4))
    x = np.linspace(xlim[0], xlim[1], 600)
    y = f(x)
    ax.plot(x, y, color="0.6", linewidth=1.5)
    a = np.asarray(points, dtype=float).ravel()
    ax.plot(a, f(a), "o-", color=couleur, markersize=4, linewidth=0.8, alpha=0.85, label=nom)
    ax.plot(a[:1], f(a[:1]), "o", color=couleur, markersize=10, markerfacecolor="none", markeredgewidth=1.8)
    ax.plot(a[-1:], f(a[-1:]), "o", color=couleur, markersize=7)
    marge = 0.08 * (np.max(y) - np.min(y))
    ax.set_xlim(xlim)
    ax.set_ylim(np.min(y) - marge, np.max(y) + marge)
    ax.set_xlabel("$a$")
    ax.grid(alpha=0.3)
    return ax


print("✔ Boîte à outils chargée : numpy (np), matplotlib (plt), tracer_iteres, tracer_contours, grad_num.")
```

## 1. La dérivée, pente de la tangente

### Sécantes et tangente 📖

La droite qui passe par deux points $p = (p_0, p_1)$ et $q = (q_0, q_1)$, avec $p_0 \neq q_0$, a pour **pente**
(coefficient directeur) $\dfrac{q_1 - p_1}{q_0 - p_0}$. Pour une fonction $f$, la droite qui passe par
$(a, f(a))$ et $(a + h, f(a + h))$ est une **sécante** au graphe ; quand $h \to 0$, sa pente tend vers la
**dérivée** :

$$f'(a) = \lim_{h \to 0} \frac{f(a + h) - f(a)}{h}.$$

La **tangente** en $a$ est la droite $y = f(a) + f'(a)\,(x - a)$ : c'est la droite qui approche le mieux le
graphe de $f$ près de $(a, f(a))$, puisque $f(a + h) = f(a) + f'(a)\, h + o(h)$.

On travaille avec la fonction $\text{fun}(x) = x \sin(x^2)$ sur $[-2, 4]$ : elle ondule de plus en plus vite.

### La dérivée numérique 📖

La pente d'une sécante approche $f'(a)$ avec une erreur proportionnelle au pas. La **différence centrée**,
pente de la sécante entre $a - h$ et $a + h$, est bien meilleure (TP 6, `grad_num`) :

$$f'(a) \approx \frac{f(a + h) - f(a - h)}{2h}, \qquad \text{erreur} \approx \frac{f'''(a)}{6}\, h^2 .$$

### La méthode du gradient 📖

Meilleure idée : un pas **proportionnel à la pente**, grand là où la pente est forte, petit près d'un minimum, où
$f'$ est presque nul :

$$a_{k+1} = a_k - e\, f'(a_k), \qquad e > 0 \text{ (le pas)}.$$

**Pourquoi $f$ diminue.** D'après Taylor, $f\big(a - e f'(a)\big) = f(a) - e\, f'(a)^2 + o(e)$ : si $f'(a) \neq 0$,
$f$ diminue pour $e$ assez petit. Et si la suite converge vers une limite $a^*$, alors $f'(a^*) = 0$ : un point
critique.

### Le rôle du pas 📖

Près d'un minimum $a^*$, $f'(a) \approx f''(a^*)\,(a - a^*)$ (Taylor appliqué à $f'$, avec $f'(a^*) = 0$), donc

$$a_{k+1} - a^* \approx \big(1 - e\, f''(a^*)\big)\,(a_k - a^*) :$$

à chaque itération, l'écart au minimum est multiplié par $1 - e\, f''(a^*)$. Il diminue si et seulement si
$|1 - e\, f''(a^*)| < 1$, c'est-à-dire

$$0 < e < \frac{2}{f''(a^*)} .$$

Plus $f$ est **courbée** au minimum, plus le pas doit être **petit**. Si $1 - e\, f''(a^*) < 0$, les points
oscillent de part et d'autre de $a^*$.

## 4. Le moment

### Donner de l'élan à la descente 📖

Le pas est délicat : trop petit, la méthode est très lente ; trop grand, elle rebondit (exercices 5 et 6). Une
autre amélioration est le **moment** (méthode de la « boule pesante », Polyak, 1964). La méthode garde la mémoire
des pas précédents, comme une boule qui roule et prend de l'élan :

$$M_0 = 0, \qquad M_{k+1} = b\, M_k + e\, f'(a_k), \qquad a_{k+1} = a_k - M_{k+1}, \qquad 0 \leq b < 1 .$$

Avec $b = 0$, c'est la méthode du gradient. Avec $b$ proche de 1, les pas qui vont dans le même sens
s'accumulent.

## 5. ★ Deux paramètres

📖 Avec une ordonnée à l'origine, la droite est $y = a\, x + b$, et l'action dépend de **deux** paramètres :

$$S(a, b) = \sum_{i=1}^{N} \big(y_i - a\, x_i - b\big)^2 .$$

### À retenir

- La **dérivée** $f'(a)$ est la pente de la tangente, limite des pentes des sécantes. Numériquement, la
  différence centrée $\frac{f(a + h) - f(a - h)}{2h}$ a une erreur en $h^2$.
- Pour descendre, on va dans le sens **opposé** au signe de $f'(a)$. Avec des pas de longueur fixe (gradient
  brutal), on oscille autour du minimum sans l'atteindre.
- **Méthode du gradient** : $a_{k+1} = a_k - e\, f'(a_k)$ ; les pas raccourcissent près du minimum. Pour $e$
  petit, $f$ diminue : $f(a - e f'(a)) = f(a) - e\, f'(a)^2 + o(e)$.
- **Le pas** : près d'un minimum $a^*$, l'écart est multiplié par $1 - e\, f''(a^*)$ à chaque itération ;
  convergence si $0 < e < 2/f''(a^*)$. Trop petit : lent ; trop grand : oscillations ou divergence.
- La limite dépend du point de départ : un minimum **local**, pas forcément le plus bas. La méthode peut aussi
  ralentir près d'un point critique qui n'est pas un minimum.
- **Moindres carrés** : la droite $y = a\, x$ qui minimise $S(a) = \sum (y_i - a\, x_i)^2$ a pour pente
  $a^* = \sum x_i y_i / \sum x_i^2$.
- **Moment** : $M_{k+1} = b\, M_k + e\, f'(a_k)$, $a_{k+1} = a_k - M_{k+1}$ : l'élan accumulé traverse les zones
  plates.
