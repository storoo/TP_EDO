### Rappels 📖

On considère le problème de Cauchy

$$y'(t) = f\big(t, y(t)\big), \quad t \in [t_0, T], \qquad y(t_0) = y_0.$$

On choisit un nombre de pas $N$ et on découpe $[t_0, T]$ en $N$ intervalles de longueur $h = \dfrac{T - t_0}{N}$
(le **pas**) : les instants de la grille sont $t_i = t_0 + i\,h$, pour $i = 0, 1, \dots, N$.

**Méthode d'Euler.** On calcule des valeurs approchées $w_i \approx y(t_i)$ par

$$\boxed{\; w_0 = y_0, \qquad w_{i+1} = w_i + h\, f(t_i, w_i), \quad i = 0, 1, \dots, N - 1. \;}$$

*Idée* : la solution qui passe par le point $(t_i, w_i)$ y a pour pente $f(t_i, w_i)$ ; on suit cette tangente
pendant un temps $h$. De façon équivalente, on néglige le reste dans la formule de Taylor
$y(t_{i+1}) = y(t_i) + h\,y'(t_i) + \frac{h^2}{2}\,y''(\xi_i)$.

**Théorème (erreur de la méthode d'Euler).** Si $f$ est $L$-lipschitzienne par rapport à $y$ et si
$|y''(t)| \leq M$ sur $[t_0, T]$, alors pour tout $i$ :

$$|y(t_i) - w_i| \leq \frac{hM}{2L}\left(e^{L(t_i - t_0)} - 1\right).$$

L'erreur est donc au plus proportionnelle à $h$ : on dit que la méthode est **d'ordre 1**.

**Exemple fil rouge** (celui de la présentation) :

$$y' = y - t^2 + 1, \quad t \in [0, 2], \quad y(0) = 0{,}5, \qquad \text{de solution exacte }
y(t) = (t + 1)^2 - \tfrac{1}{2}\, e^{t}.$$

## 3. Ordre de convergence

Si l'erreur maximale se comporte comme $E(h) \approx C\,h^p$, alors $\log E(h) \approx \log C + p \log h$ : en
échelle **log-log**, les points $(h, E(h))$ sont alignés sur une **droite de pente $p$**. On dit que la méthode est
**d'ordre $p$**. En divisant $h$ par 2, on obtient une estimation de $p$ :

$$p \approx \frac{\ln\big(E(h) / E(h/2)\big)}{\ln 2}.$$

## 4. La méthode d'Euler modifiée

### Rappel 📖

La méthode d'Euler n'utilise que la pente **au début** du pas. La méthode d'**Euler modifiée** fait la
**moyenne** de la pente au début et d'une estimation de la pente à la fin du pas :

$$
\begin{aligned}
k_1 &= f(t_i, w_i) && \text{(pente au début du pas)}\\
\tilde w_{i+1} &= w_i + h\,k_1 && \text{(prédiction par un pas d'Euler)}\\
k_2 &= f(t_{i+1}, \tilde w_{i+1}) && \text{(pente à la fin du pas, au point prédit)}\\
w_{i+1} &= w_i + \tfrac{h}{2}\,(k_1 + k_2) && \text{(correction : moyenne des deux pentes)}
\end{aligned}
$$

C'est la méthode des trapèzes,
$\displaystyle\int_{t_i}^{t_{i+1}} f\big(s, y(s)\big)\,ds \approx \frac{h}{2}\Big[f\big(t_i, y(t_i)\big) + f\big(t_{i+1}, y(t_{i+1})\big)\Big]$,
dans laquelle la valeur inconnue $y(t_{i+1})$ est **prédite** par un pas d'Euler. Chaque pas coûte deux évaluations
de $f$, mais la méthode est **d'ordre 2** : l'erreur est majorée par $C\,h^2$.

> ⚠️ **Attention aux noms.** Cette méthode s'appelle aussi **méthode de Heun** ou « Euler amélioré ». Certains
> livres appellent « Euler modifiée » une autre méthode d'ordre 2, celle du **point milieu** :
> $w_{i+1} = w_i + h\,f\big(t_i + \frac{h}{2},\, w_i + \frac{h}{2} f(t_i, w_i)\big)$.

## 7. ★ Systèmes : l'oscillateur harmonique

La méthode d'Euler s'applique **telle quelle** aux systèmes $Y' = F(t, Y)$ avec $Y(t) \in \mathbb{R}^m$ :
$W_{i+1} = W_i + h\,F(t_i, W_i)$, où les $W_i$ sont maintenant des vecteurs.

L'équation $x'' + x = 0$ (ressort sans frottement) s'écrit, avec $Y = (x, v)$ et $v = x'$ :

$$x' = v, \qquad v' = -x.$$

Pour $Y(0) = (1, 0)$, la solution est $x(t) = \cos t$, $v(t) = -\sin t$ : dans le plan $(x, v)$, elle parcourt le
**cercle** unité, et son énergie $x^2 + v^2 = 1$ reste constante.

## 8. Aide-mémoire

| Besoin | Commande |
|:--|:--|
| Grille de $[t_0, T]$ à $N$ pas | `t = np.linspace(t0, T, N + 1)` ; `h = (T - t0) / N` |
| Tableau à remplir | `w = np.zeros(N + 1)` ; `w[0] = y0` ; boucle `for i in range(N):` |
| Tableau pour un système de $m$ équations | `W = np.zeros((N + 1, m))` ; la ligne `W[i]` contient le vecteur $W_i$ |
| Valeur absolue, maximum (sans boucle) | `np.abs(...)`, `np.max(...)` |
| Graphe log-log | `ax.loglog(h_liste, E_liste, "o-")` |
| Rapports de valeurs successives | `E[:-1] / E[1:]` |
| Premier instant où une condition est vraie | `t[w > 1e6][0]` |

### À retenir

- **Euler** : $w_{i+1} = w_i + h\,f(t_i, w_i)$, une évaluation de $f$ par pas, **ordre 1** (erreur $\approx C\,h$).
- **Euler modifiée** : moyenne de deux pentes, deux évaluations de $f$ par pas, **ordre 2** (erreur $\approx C\,h^2$).
- L'ordre se lit comme une **pente** en échelle log-log.
- Le pas doit être assez petit pour que le schéma soit **stable**, et le calcul numérique ne dispense pas de
  l'analyse (explosion en temps fini, conservation de l'énergie…).
