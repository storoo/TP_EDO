```python
# 🧰 Boîte à outils du TP 6 — exécutée automatiquement à l'ouverture (sinon : Maj + Entrée).
# tracer_contours(f, xlim, ylim, ax=ax) trace les lignes de niveau de f : R^2 -> R et renvoie ax.
import math
import numpy as np
import matplotlib.pyplot as plt

def tracer_contours(f, xlim=(-2, 2), ylim=(-2, 2), n=200, niveaux=20, ax=None):
    """Trace les lignes de niveau de f : R^2 -> R sur le rectangle xlim × ylim et renvoie ax.

    f prend un point x = np.array([x, y]) et renvoie un nombre. niveaux : nombre de lignes de niveau
    (réparties selon les valeurs de f) ou liste des valeurs voulues."""
    if ax is None:
        fig, ax = plt.subplots(figsize=(5.5, 5))
    X, Y = np.meshgrid(np.linspace(xlim[0], xlim[1], n), np.linspace(ylim[0], ylim[1], n))
    try:
        Z = np.asarray(f(np.array([X, Y])), dtype=float)        # rapide si f est écrite avec x[0] et x[1]
        if Z.shape != X.shape:
            raise ValueError
    except Exception:                                           # sinon, point par point
        Z = np.array([[f(np.array([a, b])) for a, b in zip(ligne_x, ligne_y)]
                      for ligne_x, ligne_y in zip(X, Y)], dtype=float)
    if np.isscalar(niveaux):
        niveaux = np.unique(np.quantile(Z, np.linspace(0, 1, int(niveaux) + 2)[1:-1]))
    ax.contour(X, Y, Z, levels=niveaux, cmap="viridis", linewidths=1)
    ax.set_xlabel("$x$")
    ax.set_ylabel("$y$")
    ax.set_aspect("equal")
    return ax


print("✔ Boîte à outils chargée : numpy (np), matplotlib (plt), tracer_contours.")
```

### Rappels : gradient, hessienne, formule de Taylor 📖

Soit $f : \mathbb{R}^n \to \mathbb{R}$ de classe $\mathcal{C}^2$ et $x \in \mathbb{R}^n$.

- **Gradient** : $\nabla f(x) = \big(\partial_1 f(x), \dots, \partial_n f(x)\big) \in \mathbb{R}^n$.
  **Hessienne** : $\nabla^2 f(x) = \big(\partial_i \partial_j f(x)\big)_{1 \leq i, j \leq n}$, matrice
  **symétrique** $n \times n$ (théorème de Schwarz).
- **Taylor à l'ordre 2** : $f(x + h) = f(x) + \nabla f(x)^\top h + \frac{1}{2}\, h^\top \nabla^2 f(x)\, h
  + o(\|h\|^2)$.
- **Théorème spectral** : une matrice symétrique $H$ a $n$ valeurs propres **réelles**
  $\lambda_1 \leq \dots \leq \lambda_n$, avec une base orthonormée de vecteurs propres, et
  $\lambda_1 \|h\|^2 \leq h^\top H h \leq \lambda_n \|h\|^2$. $H$ est **définie positive** si $\lambda_1 > 0$,
  **semi-définie positive** si $\lambda_1 \geq 0$.

*Exemple* : $f(x, y) = x^2 + 3y^2$ a pour gradient $(2x, 6y)$ et pour hessienne
$\begin{pmatrix} 2 & 0 \\ 0 & 6 \end{pmatrix}$, définie positive.

### Minima et points critiques 📖

- $x^*$ est un **minimum local** de $f$ si $f(x^*) \leq f(x)$ pour tout $x$ proche de $x^*$ ; **global** si
  c'est vrai pour tout $x$ ; **strict** si l'inégalité est stricte pour $x \neq x^*$.
- **Condition d'ordre 1** : si $x^*$ est un minimum local, alors $\nabla f(x^*) = 0$. On dit que $x^*$ est un
  **point critique** (ou stationnaire). La réciproque est fausse.
- **Conditions d'ordre 2**, en un point critique $x^*$ : si $\nabla^2 f(x^*)$ est définie positive, $x^*$ est
  un minimum local strict ; définie négative, un maximum local strict ; si elle a une valeur propre $> 0$ et une
  $< 0$, $x^*$ est un **point selle** (ni minimum, ni maximum). Sinon, **on ne peut pas conclure**.
- **Existence** *(admis)* : si $f$ est continue et **coercive** ($f(x) \to +\infty$ quand $\|x\| \to \infty$),
  $f$ a un minimum global. **Convexité** *(admis)* : si $f$ est convexe, tout point critique est un minimum
  global.

## 6. ★ Recherche exhaustive contre algorithme

L'idée la plus simple pour minimiser $f$ sur le cube $[-2, 2]^d$ : l'évaluer sur une grille de $N$ points par
axe et garder la plus petite valeur. Avec $N = 400$, le pas de la grille est d'environ $10^{-2}$. Voici une
« boîte noire » de $d = 10$ variables : on ne connaît que ses valeurs 📖.

```python
def fabriquer_boite_noire():
    rng = np.random.default_rng(2026)
    centre = np.round(rng.uniform(-1.5, 1.5, 10), 2)        # le minimiseur, caché dans la boîte
    poids = np.arange(1, 11) / 2

    def boite_noire(x):
        """Fonction de 10 variables : on ne connaît que ses valeurs."""
        u = np.asarray(x, dtype=float) - centre
        return float(np.sum(poids * u**2) + 0.5 * np.sum(u)**2 + np.sum(u**4))

    return boite_noire


boite_noire = fabriquer_boite_noire()
print(boite_noire(np.zeros(10)))
```

### À retenir

- Un minimum local est un **point critique** ($\nabla f = 0$), mais la réciproque est fausse.
- Nature d'un point critique d'après les **valeurs propres de la hessienne** : toutes $> 0$ → minimum local
  strict ; toutes $< 0$ → maximum local strict ; de signes opposés → point selle ; une valeur propre nulle →
  on ne peut pas conclure ($f_4$ et $g$). Sur ordinateur, « nulle » veut dire « plus petite que `tol` ».
- Quadratique $\frac{1}{2} x^\top A x - b^\top x$ : gradient $Ax - b$, hessienne $A$. Si $A$ est définie
  positive, l'unique minimiseur est la solution de $Ax = b$.
- **Conditionnement** $\kappa = \lambda_{\max} / \lambda_{\min}$ : les lignes de niveau sont des ellipses de
  rapport d'axes $\sqrt{\kappa}$.
- On vérifie un gradient avec des **différences centrées** ($h \approx 10^{-6}$, erreur en $O(h^2)$).
- La recherche exhaustive coûte $N^d$ évaluations : il faut des algorithmes qui utilisent le gradient.

### Et au prochain TP…

La **descente de gradient** : partir d'un point $x_0$ et répéter $x_{k+1} = x_k - t\, \nabla f(x_k)$. C'est
exactement la méthode d'Euler (TP 2) appliquée à l'équation différentielle $x'(s) = -\nabla f\big(x(s)\big)$ :
la condition de stabilité de la méthode d'Euler deviendra une condition sur le pas $t$, et le conditionnement
$\kappa$ fixera la vitesse de convergence.
