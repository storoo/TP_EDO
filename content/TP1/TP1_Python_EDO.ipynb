## 1. Bases de Python

### 1.1 Nombres, variables, opérations 📖

````python
a = 7
b = 2
print(a + b, a - b, a * b, a / b)   # la division / donne toujours un réel
print(a // b, a % b)                # quotient et reste de la division euclidienne
print(a ** b)                       # puissance : a^b s'écrit a ** b en Python
````

> ⚠️ **Piège classique :** en Python, `a ^ b` n'est **pas** une puissance (c'est un « ou exclusif » bit à bit).
> Écrivez toujours `a ** b`.

Les réels sont stockés en **virgule flottante**, avec une précision finie (environ 16 chiffres). C'est essentiel en
calcul numérique : on ne compare jamais deux réels avec `==`, mais on regarde si leur écart est petit.

## 4. Champs de directions et analyse qualitative

### 4.1 Le principe

Pour l'EDO $y' = f(t, y)$, une solution qui passe par le point $(t, y)$ y a pour pente $f(t, y)$. Le **champ de
directions** consiste à tracer, en chaque point d'une grille, un petit segment de pente $f(t, y)$, c'est-à-dire
dirigé par le vecteur $(1, f(t, y))$. Les courbes solutions sont **tangentes** à ces segments : sans résoudre
l'équation, on « voit » le comportement des solutions.

### 4.2 Grilles 2D avec `np.meshgrid` 📖

`np.meshgrid` transforme deux tableaux 1D en deux tableaux 2D `T` et `Y` : le point numéro $(i, j)$ de la grille a
pour coordonnées `(T[i, j], Y[i, j])`.

````python
T, Y = np.meshgrid(np.linspace(0, 3, 4), np.linspace(-1, 1, 3))
print("T =\n", T)
print("Y =\n", Y)
print("pentes de y' = y - t :\n", Y - T)     # calcul vectorisé en tous les points de la grille
````

### 4.3 Tracer le champ avec `quiver` 📖

`ax.quiver(T, Y, U, V)` trace au point `(T, Y)` le vecteur `(U, V)`. On **normalise** le vecteur $(1, f)$ pour que
tous les segments aient la même taille, sinon les grandes pentes écrasent la figure. Exemple du cours :
$v' = 9{,}8 - v/5$.

````python
def f(t, v):
    return 9.8 - v / 5

T, V = np.meshgrid(np.linspace(0, 30, 22), np.linspace(0, 90, 22))
P = f(T, V)                         # pentes en chaque point
norme = np.sqrt(1 + P ** 2)         # norme du vecteur (1, P)

fig, ax = plt.subplots(figsize=(9, 5.5))
ax.quiver(T, V, 1 / norme, P / norme, angles="xy", color="0.6",
          headwidth=0, headlength=0, headaxislength=0, pivot="mid")   # segments sans pointe, centrés
ax.axhline(49, ls="--", color="crimson", label="équilibre $v = 49$")
ax.set_xlabel("$t$ (s)")
ax.set_ylabel("$v$ (m/s)")
ax.set_title(r"Champ de directions de $v' = 9{,}8 - v/5$")
ax.legend(loc="lower right")
plt.show()
````

### 5.1 Syntaxe de `solve_ivp` 📖

```python
sol = solve_ivp(f, (t0, tf), y0, t_eval=..., args=..., rtol=..., atol=...)
```

- `f(t, y)` : le second membre (toujours dans l'ordre `t, y`) ;
- `(t0, tf)` : l'intervalle de temps ;
- `y0` : la condition initiale, **toujours sous forme de liste** (même pour une EDO scalaire) : `[0.0]` ;
- `t_eval` : les instants où l'on veut la solution ; `args` : les paramètres supplémentaires de `f` ;
- `rtol`, `atol` : les tolérances d'erreur (par défaut `1e-3` et `1e-6`).

Le résultat `sol` contient `sol.t` (les instants) et `sol.y`, un tableau de forme `(dimension, nombre d'instants)` :
la solution d'une EDO scalaire est donc `sol.y[0]`.

## 6. ★ Pour aller plus loin : systèmes d'EDO et portraits de phase

Pour un système $Y' = F(t, Y)$ avec $Y \in \mathbb{R}^n$, `F(t, Y)` reçoit un tableau `Y` de taille $n$ et renvoie un
tableau de taille $n$. La condition initiale `y0` est une liste de $n$ nombres, et `sol.y` a $n$ lignes.

### 6.1 Le modèle proie–prédateur de Lotka–Volterra 📖

$$x' = \alpha x - \beta x y, \qquad y' = -\gamma y + \delta x y,$$

où $x$ désigne les proies et $y$ les prédateurs. Le **portrait de phase** représente les trajectoires
$t \mapsto (x(t), y(t))$ dans le plan ; `streamplot` trace des lignes de champ.

````python
alpha, beta, gamma, delta = 1.0, 0.1, 1.5, 0.075

def lotka_volterra(t, Y):
    x, y = Y
    return np.array([alpha * x - beta * x * y,
                     -gamma * y + delta * x * y])

t_eval = np.linspace(0, 15, 1000)
sol = solve_ivp(lotka_volterra, (0, 15), [10, 5], t_eval=t_eval, rtol=1e-8)
x, y = sol.y                                    # sol.y a la forme (2, 1000)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5), layout="constrained")
ax1.plot(sol.t, x, label="proies $x(t)$")
ax1.plot(sol.t, y, label="prédateurs $y(t)$")
ax1.set_xlabel("$t$")
ax1.legend()
ax1.grid(alpha=0.3)

X, Y = np.meshgrid(np.linspace(0.1, 60, 30), np.linspace(0.1, 35, 30))
U, V = lotka_volterra(0, [X, Y])                # le champ de vecteurs, calculé sur toute la grille
ax2.streamplot(X, Y, U, V, color="0.75", density=1.2)
for x0 in [5, 10, 15]:
    s = solve_ivp(lotka_volterra, (0, 15), [x0, 5], t_eval=t_eval, rtol=1e-8)
    ax2.plot(s.y[0], s.y[1], lw=2, label=f"$(x_0, y_0) = ({x0}, 5)$")
ax2.plot(gamma / delta, alpha / beta, "k*", ms=12, label="équilibre")
ax2.set_xlabel("proies $x$")
ax2.set_ylabel("prédateurs $y$")
ax2.legend(loc="upper right")
plt.show()
````

## 7. ★ Pour aller plus loin : calcul symbolique avec SymPy

**SymPy** manipule des expressions mathématiques **exactes** : dérivées, simplifications, limites, et même
résolution de certaines EDO (`dsolve`). C'est pratique pour **vérifier** un calcul à la main, ou pour obtenir la
solution exacte qui servira à mesurer l'erreur d'une méthode numérique.

> On nomme les objets symboliques `ts`, `rs`, `Ks`… (« s » pour symbolique), pour ne pas écraser les tableaux
> NumPy `t`, `N`, etc. utilisés plus haut.

````python
import sympy as sp

ts = sp.symbols("t")
rs, Ks, N0s = sp.symbols("r K N_0", positive=True)
Ns = sp.Function("N")

# Modèle de Malthus : dsolve résout le problème de Cauchy
edo_malthus = sp.Eq(Ns(ts).diff(ts), rs * Ns(ts))
display(edo_malthus)
display(sp.dsolve(edo_malthus, Ns(ts), ics={Ns(0): N0s}))
````

````python
# Vérifier la solution de Verhulst donnée en cours
N_verhulst = Ks * N0s * sp.exp(rs * ts) / (Ks + N0s * (sp.exp(rs * ts) - 1))
residu = sp.diff(N_verhulst, ts) - rs * N_verhulst * (1 - N_verhulst / Ks)

print("résidu simplifié :", sp.simplify(residu))                  # 0 : c'est bien une solution
print("valeur en t = 0  :", sp.simplify(N_verhulst.subs(ts, 0)))  # N_0
print("limite en +oo    :", sp.limit(N_verhulst, ts, sp.oo))       # K

# lambdify : transformer l'expression en fonction NumPy, pour la tracer
N_num = sp.lambdify(ts, N_verhulst.subs({rs: 1, Ks: 1, N0s: 0.1}), "numpy")
print(N_num(np.linspace(0, 10, 6)))
````
